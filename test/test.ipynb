```python
# 1. Instala la librería si no la tienes (ejecuta esto en una celda)
# !pip install qrcode[pil]

import qrcode
from IPython.display import display

# 2. El texto exacto que quieres que lea tu app (La placa del bus)
texto_placa = "ABI713"

# 3. Configuración del QR
qr = qrcode.QRCode(
    version=1, # Tamaño del QR (1 es el más pequeño)
    error_correction=qrcode.constants.ERROR_CORRECT_H, # Alta corrección de errores (ideal para escanear de lejos)
    box_size=10, # Tamaño de cada "cuadrito"
    border=4, # Margen blanco alrededor
)

# 4. Agregamos el texto y generamos
qr.add_data(texto_placa)
qr.make(fit=True)

# 5. Creamos la imagen (Blanco y Negro)
img = qr.make_image(fill_color="black", back_color="white")

# 6. Guardamos el archivo localmente en la carpeta del notebook
img.save(texto_placa + ".png")

# 7. Lo mostramos directamente en el output del Notebook
display(img)
```
