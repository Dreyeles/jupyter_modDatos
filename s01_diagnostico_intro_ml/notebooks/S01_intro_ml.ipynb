# S01 — Introducción práctica a ML

**Caso práctico:** Clasificación básica con Iris Dataset

**Objetivo:**
- Cargar un dataset simple (Iris).
- Entrenar un modelo de clasificación (Árbol de decisión).
- Predecir y mostrar la precisión.


```python
# 1. Importar librerías
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python

# 2. Cargar dataset
iris = load_iris()
X = iris.data
y = iris.target
```

```python

# 3. Dividir en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python

# 4. Crear y entrenar modelo
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
```

```python

# 5. Predecir y evaluar
y_pred = model.predict(X_test)
print("Precisión del modelo:", accuracy_score(y_test, y_pred))
```
