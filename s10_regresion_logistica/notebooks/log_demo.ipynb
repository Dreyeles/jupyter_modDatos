```python
import numpy as np
import matplotlib.pyplot as plt

# Datos ficticios
x = np.linspace(-10, 10, 200)
# Simula una variable dependiente binaria con ruido
y = (x > 0).astype(int) + np.random.normal(0, 0.1, x.shape)

# Línea recta (regresión lineal)
y_lineal = 0.1 * x + 0.5

# Curva S (regresión logística)
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

y_logistica = sigmoid(x)

plt.figure(figsize=(8,5))
plt.scatter(x, y, color='gray', alpha=0.5, label='Datos')
plt.plot(x, y_lineal, color='red', linestyle='--', label='Regresión lineal')
plt.plot(x, y_logistica, color='blue', linewidth=2, label='Regresión logística (curva S)')
plt.ylim(-0.5, 1.5)
plt.xlabel('X')
plt.ylabel('Probabilidad / Valor')
plt.title('Regresión Lineal vs Regresión Logística')
plt.legend()
plt.grid(True)
plt.show()
```
