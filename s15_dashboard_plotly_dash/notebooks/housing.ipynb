```python
import pandas as pd
import os

# 1. Cargar el dataset original (housing.csv)

df = pd.read_csv("../datasets/housing.csv")

print(f"Original: {df.shape} filas")

# 2. EL PROCESAMIENTO (Limpieza básica)
df_clean = df.dropna()

# Opcional: La guía a veces pide convertir 'ocean_proximity' a números (Get Dummies).
df_clean = df_clean.select_dtypes(include=['float64', 'int64'])

print(f"Procesada: {df_clean.shape} filas (Nulos eliminados)")

# 3. GUARDAR EL ARCHIVO (Creando la carpeta 'data' si no existe)
if not os.path.exists('../data'):
    os.makedirs('../data')

# Guardamos el CSV. index=True es importante porque la guía usa index_col=0 al leerlo
df_clean.to_csv('../data/data_procesada.csv', index=True)

print("\n ¡Listo! Archivo generado en: data/data_procesada.csv")
```
