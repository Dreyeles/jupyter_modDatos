```python
# Datos de la imagen
x = [7, 1, 10, 5, 4, 3, 13, 10, 2]
y = [2, 9, 2, 5, 7, 11, 2, 5, 14]

import numpy as np

# Convertir a arrays de numpy
x = np.array(x)
y = np.array(y)

# Calcular los coeficientes m y b
m, b = np.polyfit(x, y, 1)

print(f"La ecuación de la recta es: y = {m:.2f}x + {b:.2f}")
```

```python
n = len(x)
sum_x = np.sum(x)
sum_y = np.sum(y)
sum_xy = np.sum(x * y)
sum_x2 = np.sum(x ** 2)

m = (sum_xy - (sum_x * sum_y) / n) / (sum_x2 - (sum_x ** 2) / n)
b = (sum_y / n) - m * (sum_x / n)

print(f"La ecuación de la recta es: y = {m:.2f}x + {b:.2f}")
```

```python
import matplotlib.pyplot as plt

# Crear una figura solo para mostrar las fórmulas y resultados
fig, ax = plt.subplots(figsize=(7, 3))
ax.axis('off')

# Texto con las fórmulas y resultados usando LaTeX
formula = (
    r"$\text{Fórmulas:}$" "\n"
    r"$y = mx + b$" "\n"
    r"$m = \frac{\sum xy - \frac{\sum x \sum y}{n}}{\sum x^2 - \frac{(\sum x)^2}{n}}$" "\n"
    r"$b = \bar{y} - m \bar{x}$" "\n\n"
    r"$\text{Resultados:}$" "\n"
    fr"$m = {m:.2f}$" "\n"
    fr"$b = {b:.2f}$" "\n"
    fr"$y = {m:.2f}x + {b:.2f}$"
)

ax.text(0.5, 0.5, formula, fontsize=16, ha='center', va='center')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Calcular el error RMSE
y_pred = m * x + b
rmse = np.sqrt(np.mean((y - y_pred) ** 2))

# Mostrar la fórmula y el resultado del RMSE en gráfico
fig, ax = plt.subplots(figsize=(7, 2.5))
ax.axis('off')

rmse_text = (
    r"$\text{Error cuadrático medio (RMSE):}$" "\n"
    r"$RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$" "\n\n"
    fr"$RMSE = {rmse:.2f}$"
)

ax.text(0.5, 0.5, rmse_text, fontsize=16, ha='center', va='center')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Calcular el coeficiente de determinación R²
y_pred = m * x + b
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r2 = 1 - (ss_res / ss_tot)

# Mostrar la fórmula y el resultado de R² en gráfico
fig, ax = plt.subplots(figsize=(7, 2.5))
ax.axis('off')

r2_text = (
    r"$\text{Coeficiente de determinación (R^2):}$" "\n"
    r"$R^2 = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}$" "\n\n"
    fr"$R^2 = {r2:.2f}$"
)

ax.text(0.5, 0.5, r2_text, fontsize=16, ha='center', va='center')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Calcular el error MAE
y_pred = m * x + b
mae = np.mean(np.abs(y - y_pred))

# Mostrar la fórmula y el resultado del MAE en gráfico
fig, ax = plt.subplots(figsize=(7, 2.5))
ax.axis('off')

mae_text = (
    r"$\text{Error absoluto medio (MAE):}$" "\n"
    r"$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$" "\n\n"
    fr"$MAE = {mae:.2f}$"
)

ax.text(0.5, 0.5, mae_text, fontsize=16, ha='center', va='center')

plt.show()
```
