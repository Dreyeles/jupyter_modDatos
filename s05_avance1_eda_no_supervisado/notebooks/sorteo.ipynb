```python
import random

# --- 1. Definir la lista de los 11 grupos ---
grupos = [
    "Grupo 1",
    "Grupo 2",
    "Grupo 3",
    "Grupo 4",
    "Grupo 5"
]

print(f"Lista original de grupos: {grupos}\n")

# --- 2. Revolver la lista (aleatorizar) ---
# random.shuffle() modifica la lista en su lugar (in-place).
random.shuffle(grupos)

# --- 3. Imprimir el orden de exposición ---
print("--- ORDEN DE EXPOSICIÓN (AVANCE 1) ---")
print("Sorteo realizado. El orden es:\n")

for i, grupo in enumerate(grupos):
    print(f"Turno {i+1}: {grupo}")
```
