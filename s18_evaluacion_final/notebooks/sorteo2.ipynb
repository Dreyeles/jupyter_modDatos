```python
import random

grupos = [
    "Grupo 1",
    "Grupo 2",
    "Grupo 3",
    "Grupo 4"
]

print(f"Lista original de grupos: {grupos}\n")

random.shuffle(grupos)

print("--- ORDEN DE EXPOSICIÓN (Final) ---")
print("Sorteo realizado. El orden es:\n")

for i, grupo in enumerate(grupos):
    print(f"Turno {i+1}: {grupo}")
```
